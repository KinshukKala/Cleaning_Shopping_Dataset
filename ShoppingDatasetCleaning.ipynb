Here we see that following columns contain missing values:
discount(121),
what_customer_said(573),
seller_name(301),,
videos(781),
seller_information(301),
variations(562).

discount: discount is a numeric column and contains few missing values (121 only) so I will replace the missing values with the median of the column.

```python
df['discount'].fillna(df['discount'].median(), inplace=True)
```

seller_name & seller_information: Both contains 301 missing values and maybe useful later so I will fill missing values for these columns with NA

```python
df['seller_name'].fillna("NA", inplace=True)
df['seller_information'].fillna("NA", inplace=True)
```

what_customer_said & variations: both contain nearly half missing values. what_customer_said maybe an important feature for tasks like sentiment analysis hence I am keeping it. variations may not be an important feature and because it contains lots of missing values, I am removing it.

```python
df['what_customers_said']=df['what_customers_said'].fillna("NA")
df.drop(columns=['variations'], inplace=True)
```

videos: videos has 781 missing values which is a lot so I am dropping it.

```python
df.drop(columns=['videos'], inplace=True)
```

```python
df.isnull().sum()
```

Now I will create a new data frame with useful features. Here are the features that I am not using in the new dataset:

1.   image: it contains data in the form of url which is not ideal for ml model
2.   amount_of_stars : the information provided by this column is already provided by ratings. Hence it is a redundant column and is removed.
3.   breadcrumbs: the information provided by this column is already provided by category column. Hence it is a redundant column and is removed.It also contains urls
4.   best_offer: it mostly contains {} empty values and other rows give information already provided by discount and final_price
5. more_offers : contain information already given by discount and final_price hence removed,






```python
useful_features = [
    "product_id",
    "title",
    "product_description",
    "rating",
    "ratings_count",
    "initial_price",
    "discount",
    "final_price",
    "currency",
    "delivery_options",
    "product_details",
    "product_specifications",
    "what_customers_said",
    "seller_name",
    "sizes",
    "seller_information",
    "category"
]

df_new = df[useful_features].copy()
```

Here are some row filtered datasets:
1.) high_rated : contains products with ratings more than 4.5. Useful for extracting highly rated products.
2.) high_discount : contains products with discount more than 50%. Useful for knowing which products have high discounts on them.

```python
high_rated = df_new[df_new['rating'] > 4.5]
high_discount = df_new[df_new['discount'] > 50]
```

Creating a derived column : total_amount = price * quantity\
Since our dataset doesnt have a quantity column I will create a new column quantity filled with random values.\



```python
import numpy as np
np.random.seed(42)
df_new['quantity']=np.random.randint(1,10,size=len(df_new))
```

Also, final_price datatype is object so I will convert it to float.

```python
df_new["final_price"] = (df_new["final_price"].str.replace("₹", "", regex=False))
df_new["final_price"] = (df_new["final_price"].str.replace(",", "", regex=False))
df_new["final_price"] = (df_new["final_price"].str.replace('"', "", regex=False))
df_new['final_price']=pd.to_numeric(df_new['final_price'])
```

```python
df_new['final_price'].head()
```

Finally, new feature total_amount

```python
df_new['final_price']=df_new['quantity']*df_new['final_price']
```
